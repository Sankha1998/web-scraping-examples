```python
def get_summery_data(i,j):
    url = f"https://results.eci.gov.in/ResultAcGenMay2026/Constituencywise{j}.htm"
    browser = webdriver.Chrome("./chromedriver")
    browser.get(url)
    


    soup = bs(browser.page_source,"html.parser")
    table = soup.find('table')
    if not table:
        print("No table found on the page.")


    headers = []
    header_row = table.find('tr')
    if header_row:
        headers = [th.get_text(strip=True) for th in header_row.find_all(['th', 'td'])]

    
    rows_data = []
    for row in table.find_all('tr')[1:]:  
        cols = row.find_all(['td', 'th'])
        cols_text = [col.get_text(strip=True) for col in cols]
        if cols_text:  
            rows_data.append(cols_text)

  
    headers.append('Constituency')
    
    rows_data = rows_data[:-1]
    for x in rows_data:
        t = j.replace('S22',"")
        x.append(f'{t} - {i}')
    df = pd.DataFrame(rows_data[:-1], columns=headers)
    return df
```

```python
def get_round_wise_counting(q,w):
    records=[]
    url = f"https://results.eci.gov.in/ResultAcGenMay2026/Roundwise{w}.htm"
    browser = webdriver.Chrome("./chromedriver")
    browser.get(url)
    soup = bs(browser.page_source,"html.parser")
    table = soup.find_all('div',class_='custom-table tabcontent')

    t = w.replace('S22',"")
    for round_num,i in enumerate(table):
        tbody = i.find('tbody')
        for tr in tbody.find_all("tr"):

            cells = [td.get_text(strip=True) for td in tr.find_all("td")]
           
            if len(cells) < 5:
                continue  # skip malformed / footer rows

            candidate, party, votes_prev, votes_current, votes_total = cells[:5]

            records.append({
                "round":         round_num+1,
                "candidate":     candidate,
                "party":         party,
                "votes_prev":    votes_prev.replace(",", ""),
                "votes_current": votes_current.replace(",", ""),
                "votes_total":   votes_total.replace(",", ""),
                "Constituency": f'{t} - {q}'
            })      
    df = pd.DataFrame.from_dict(records)
    return df
        
```

```python
dfs = []
c = 1
for q,w in zip(nms,numbs):
    try:
        dfs.append(get_round_wise_counting(q,w))
        print(c)
        c+=1
    except:
        print(f'error{c}')
```

```python
combined_df = pd.concat(dfs, ignore_index=True)
combined_df.shape
combined_df.to_csv('result_round_wise_counting.csv', index=False)
```
